```python
a=np.array([[10,20,30],[40,50,60]])
print ('shape:',a.shape)
print ('dimensions:',a.ndim)
print ('size:',a.size)
print ('data types:',a.dtype)
print ('rows:',a.shape[0],'columns:',a.shape[1])
```

```python
a=np.array([10,20,30,40])
print (a+5)
print (a-5)
print (a*2)
print (a/10)
print (a**2)
print (np.sum(a))
print (np.mean(a))
print (np.median(a))
print (np.min(a),np.max(a))
print (np.std(a))
```
