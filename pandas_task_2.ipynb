exercise-4 sales data analysis

```python
import pandas as pd

data = {
    'Product': ['Laptop', 'Phone', 'Tablet', 'Camera', 'Headset'],
    'Mon': [5, 10, 4, 2, 8],
    'Tue': [6, 12, 5, 3, 9],
    'Wed': [3, 11, 2, 1, 7],
    'Thu': [4, 9, 3, 2, 6],
    'Fri': [7, 13, 5, 4, 10],
    'Sat': [8, 15, 6, 5, 11],
    'Sun': [6, 14, 4, 3, 9]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)


df['Total Sales'] = df.iloc[:, 1:8].sum(axis=1)

print("\nTotal Weekly Sales:")
print(df[['Product', 'Total Sales']])


best = df.loc[df['Total Sales'].idxmax(), 'Product']
print("\nBest Selling Product:", best)


daily_average = df.iloc[:, 1:8].sum().mean()
print("\nAverage Sales Per Day:", daily_average)


df = df.sort_values(by='Total Sales', ascending=False)

print("\nSorted DataFrame:")
print(df)
```

exercise-5 employee attendance tracker


```python
import pandas as pd

data = {
    'Employee': ['John', 'Alice', 'Bob', 'Mary', 'David'],
    'Mon': ['P', 'P', 'A', 'P', 'P'],
    'Tue': ['A', 'P', 'A', 'P', 'P'],
    'Wed': ['P', 'P', 'P', 'P', 'A'],
    'Thu': ['P', 'P', 'P', 'A', 'A'],
    'Fri': ['P', 'A', 'P', 'P', 'A']
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)


df['Total_Present'] = (df.iloc[:, 1:6] == 'P').sum(axis=1)

print("\nTotal Present Days:")
print(df[['Employee', 'Total_Present']])


full = df[df['Total_Present'] == 5]
print("\nFull Attendance Employees:")
print(full['Employee'])

absentees = (df.iloc[:, 1:6] == 'A').sum()
max_absent = absentees.max()

print("\nDay with Maximum Absentees:")
print(absentees[absentees == max_absent])


df = df.sort_values(
    by='Total_Present',
    ascending=False,
    kind='stable'
)

print("\nSorted DataFrame:")
print(df)
```

exercise-6 pandas-movie ratings analysis

```python
import pandas as pd


data = {
    'Movie': ['Inception', 'Titanic', 'Avatar', 'Joker', 'Dune'],
    'Critic_A': [9, 8, 7, 9, 8],
    'Critic_B': [8, 7, 8, 9, 7],
    'Critic_C': [9, 8, 6, 10, 8]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)


df['Avg_Rating'] = df.iloc[:, 1:4].mean(axis=1)

print("\nAverage Rating:")
print(df[['Movie', 'Avg_Rating']])


highest = df.loc[df['Avg_Rating'].idxmax(), 'Movie']
print("\nHighest Rated Movie:", highest)


lowest = df.loc[df['Avg_Rating'].idxmin(), 'Movie']
print("\nLowest Rated Movie:", lowest)


df = df.sort_values(
    by='Avg_Rating',
    ascending=False,
    kind='stable'
)

print("\nSorted DataFrame:")
print(df)
```
