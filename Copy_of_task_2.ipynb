```python
year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("It is a leap year")
else:
    print("It is not a leap year")
```

```python
books = ["Python", "Java", "C", "HTML", "SQL"]

want = input("Do you want to take a book? (yes/no): ")

if want == "yes":
    card = input("Do you have a library card? (yes/no): ")

    if card == "yes":
        print("Available books:", books)

        book = input("Enter the book you want: ")

        if book in books:
            books.remove(book)
            print("Book issued successfully")
            print("Remaining books:", books)
        else:
            print("Book is not available")
    else:
        print("You need a library card to take a book")
else:
    print("Thank you")
```

```python
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```

```python
num = int(input("Enter a number: "))

original = num
sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit ** 3
    num = num // 10

if sum == original:
    print("It is an Armstrong number")
else:
    print("It is not an Armstrong number")
```

```python
def unique_list(numbers):
    result = []

    for num in numbers:
        if num not in result:
            result.append(num)

    return result


numbers = [1, 2, 3, 3, 3, 3, 4, 5]

print("Original List:", numbers)
print("Unique List:", unique_list(numbers))
```

```python
def count_vowels(string):
    count = 0

    for ch in string:
        if ch.lower() in "aeiou":
            count = count + 1

    return count


text = input("Enter a string: ")

print("Number of vowels:", count_vowels(text))
```

```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


num = int(input("Enter a number: "))

if is_prime(num):
    print("The number is prime")
else:
    print("The number is not prime")
```

```python
def calculator(a, b, operator):

    if operator == "+":
        return a + b

    elif operator == "-":
        return a - b

    elif operator == "*":
        return a * b

    elif operator == "/":
        return a / b

    else:
        return "Invalid operator"


num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

result = calculator(num1, num2, operator)

print("Result =", result)
```
