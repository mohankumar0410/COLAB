```python
numbers = [10, 15, 20, 25, 30, 35]

even = 0
odd = 0

for i in numbers:
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers =", even)
print("Odd numbers =", odd)
```

```python
numbers = [10, -5, 0, 20, -8, 0, 15]

positive = 0
negative = 0
zero = 0

for i in numbers:
    if i > 0:
        positive = positive + 1
    elif i < 0:
        negative = negative + 1
    else:
        zero = zero + 1

print("Positive =", positive)
print("Negative =", negative)
print("Zero =", zero)
```

```python
n=int(input("enter a number:"))
original=n
reverse=0
while n>0:
  digit=n%10
  reverse=reverse*10+digit
  n =n//10
if original ==reverse:
    print("palindrome")
else:
    print("not palindrome")
```
