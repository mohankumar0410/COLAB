## 2. Loops

i. Write a program to print the factorial of a number.

```python
n = int(input("Enter a number: "))
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print("Factorial:", factorial)
```

ii. Write a program to check whether the number is Armstrong or not.

```python
num = int(input("Enter a number: "))
original = num
digits = len(str(num))
total = 0

while num > 0:
    digit = num % 10
    total += digit ** digits
    num //= 10

if total == original:
    print("It is an armstrong number")
else:
    print("It is not an armstrong number")
```

iii. Write a program to print a star pattern.

```python
n = int(input("Enter number of rows: "))

for i in range(1, n + 1):
    print("* " * i)
```

iv. Write a program to print the sum of digits accepted from the user.

```python
n = int(input("Enter a number: "))
total = 0

while n > 0:
    digit = n % 10
    total += digit
    n //= 10

print("Sum of digits:", total)
```

v. Accept 10 numbers from the user and display the average of all the numbers.

```python
total = 0

for i in range(10):
    num = int(input(f"Enter number {i + 1} : "))
    total += num

average = total / 10
print("Average:", average)
```

## 3. Functions

i. Write a Python function that takes a list and returns a new list with distinct (unique) elements from the first list.

```python
def unique_elements(my_list):
    return list(dict.fromkeys(my_list))

sample_list = [1, 2, 3, 3, 3, 3, 4, 5]
print("Original list:", sample_list)
print("Unique list:", unique_elements(sample_list))
```

ii. Write a function count_vowels(string) that takes a string and returns the number of vowels (a, e, i, o, u) in the string.

```python
def count_vowels(string):
    count = 0
    for char in string.lower():
        if char in "aeiou":
            count += 1
    return count

text = input("Enter a string: ")
print("Number of vowels:", count_vowels(text))
```

iii. Create a function is prime(n) that takes a number and returns True if the number is prime, otherwise False.

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

number = int(input("Enter a number: "))
print(is_prime(number))
```

iv. Create a simple calculator function.

```python
def calculator(a, b, operator):

    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        if b == 0:
            return "Cannot divide by zero"
        return a / b
    else:
        return "Invalid operator"

operator = input("Enter the operator (+, -, *, /): ")
num1 = int(input("Enter the first number  : "))
num2 = int(input("Enter the second number : "))


print("Result:", calculator(num1, num2, operator))
```
