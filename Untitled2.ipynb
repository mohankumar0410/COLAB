```python
my_tuple = (10, 20, 30, 40, 50, 60)

slice1 = my_tuple[1:4]

slice2 = my_tuple[2:]

slice3 = my_tuple[:3]

slice4 = my_tuple[::2]

print(f"Original tuple: {my_tuple}")
print(f"Slice 1: {slice1}")
print(f"Slice 2: {slice2}")
print(f"Slice 3: {slice3}")
print(f"Slice 4: {slice4}")
```
