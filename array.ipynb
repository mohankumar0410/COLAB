array manipulation

```python
import numpy as np

a = np.arange(1,7)
print (a.reshape(2,3))
print(a.reshape(3,2))
print (a.flatten())
print (np.transpose(a.reshape(2,3)))
```

joining array

```python
a=np.array([1,2])
b=np.array([2,3])
print (np.concatenate((a,b)))
print (np.vstack((a,b)))
print (np.hstack((a,b)))
```

splitting array

```python
a=np.array([10,20,30,40,50,60])
print (np.split(a,3))
print (np.append (a, [70,80]))
print (np.insert (a,2, [100,200]))
print (np.delete (a,2))
print (np.resize (a,(2,3)))
print (np.swapaxes (a.reshape(2,3),0,1))
```
