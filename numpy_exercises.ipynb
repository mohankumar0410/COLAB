exercise-2 electricity consumption analysis


```python
import numpy as np


electricity = np.array([12.5,15.2,14.8,13.5,16.0,17.1,15.6])
print("electricity consumption:",electricity)
```

```python
print("total consumption:",np.sum(electricity))
print("average consumption:",np.mean(electricity))
```

```python
print("highest consumption day:",np.max(electricity))
print("lowest consumption day:",np.min(electricity))
```

```python
print("daily difference:",np.diff(electricity))
```

```python
print("consumption above 16:",electricity[electricity>16])
```

exercise-3 student marks analysis

```python
import numpy as np
marks=np.array([85,90,78,88,92])
print("student marks:",marks)
```

```python
average=np.mean(marks)

print("average marks:",average)
print("highest marks:",np.max(marks))
print("lowest marks:",np.min(marks))
```

```python
print("difference from average:",marks-average)
```

```python
percentage = marks.astype(float)
print("percentage marks:",percentage)
```

```python
print("marks below 80:",marks[marks<80])
```

exercise-4 stock price tracker

```python
import numpy as np

stock = np.array([102.5,104.2,101.8,103.0,105.5,107.3])
print("stock prices:",stock)
```

```python
print("mean price:",np.mean(stock))
print("maximum price:",np.max(stock))
print("minimum price:",np.min(stock))
```

```python
daily_change = np.diff(stock)
print("daily change:",daily_change)
```

```python
percentage_change=(daily_change/stock[:-1])*100
print("percentage change:",percentage_change)
```

```python
days=np.where(percentage_change>1.5)[0]+2
print("days with increase above 1.5%:",days)
```
